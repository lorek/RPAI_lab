## Model i funkcja symulująca grę
Dwaj gracze mają łącznie M zł, z czego pierwszy zaczyna z N zł. W każdej niezależnej rundzie pierwszy gracz zyskuje 1 zł z prawdopodobieństwem p lub traci 1 zł z prawdopodobieństwem 1 − p. Gra kończy się po osiągnięciu 0 lub M.

Poniższa funkcja jest narzędziem do eksperymentów z tej listy. Zwraca kapitał od chwili 0 do zakończenia gry. Przyjmujemy całkowite 0 < N < M i 0 < p < 1.

```python
def play_game(N, M, p, rng):
    capital = N
    path = [capital]
    while 0 < capital < M:
        if rng.random() < p:
            capital += 1
        else:
            capital -= 1
        path.append(capital)
    return np.array(path)
```
