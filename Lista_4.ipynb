## 5. Średnia liczba produktów i średnie warunkowe
X oznacza liczbę produktów, a Y sposób dostawy: 0 -- paczkomat, 1 -- kurier.
Wartość oczekiwana liczby produktów jest średnią ważoną możliwych wartości X:

$$E[X]=\sum_{x=1}^{3}x\,P(X=x).$$

Jeśli ograniczamy się do zamówień z ustalonym sposobem dostawy y, używamy odpowiedniego rozkładu warunkowego:

$$E[X\mid Y=y]=\sum_{x=1}^{3}x\,P(X=x\mid Y=y),\qquad
P(X=x\mid Y=y)=\frac{P(X=x,Y=y)}{P(Y=y)}.$$
