## 1. Dwie kostki: wybieramy wyniki spełniające warunek
Rzucamy dwiema niezależnymi, uczciwymi kostkami. Każdy wiersz tablicy opisuje jeden rzut parą kostek.

Przed symulacją porównaj trzy pytania:
1. Jaka jest szansa szóstki na pierwszej kostce?
2. Jaka jest ta szansa, jeśli suma oczek wynosi co najmniej 10?
3. Jaka jest ta szansa, jeśli suma oczek wynosi dokładnie 8?

```python
n = 100_000
dice = rng.integers(1, 7, size=(n, 2))
totals = dice.sum(axis=1)
first_six = dice[:, 0] == 6
large_sum = totals >= 10
print("Bez warunku:", first_six.mean())
print("Liczba rzutów z sumą co najmniej 10:", large_sum.sum())
if large_sum.any():
    print("Wśród rzutów z sumą co najmniej 10:", first_six[large_sum].mean())
```

## 2. Dziesięć kostek: jedynki, dwójki i wyniki nieparzyste
Rzucamy dziesięcioma niezależnymi, uczciwymi kostkami. Czy informacja, że co najmniej siedem wyników jest nieparzystych, zwiększa szansę uzyskania większej liczby jedynek niż dwójek?
Zapisz przewidywanie i uzasadnienie przed uruchomieniem symulacji.

```python
ten_dice = rng.integers(1, 7, size=(100_000, 10))
odd_counts = (ten_dice % 2 == 1).sum(axis=1)
print("Pierwsze trzy rzuty:\n", ten_dice[:3])
print("Liczby wyników nieparzystych:", odd_counts[:3])
```

## 3. Dodatni test — czy to znaczy, że jestem chory?
Test daje dodatni wynik u 92% chorych i u 6% zdrowych. W rozważanej populacji choruje 10% osób.

**Przewidywanie:** spośród osób z dodatnim wynikiem, jaki odsetek jest chory? Czy to 92%?

Jedna osoba = jeden wiersz. Najpierw losujemy stan zdrowia, następnie wynik testu z prawdopodobieństwem zależnym od tego stanu.

```python
n_people = 100_000
prevalence = 0.10
sensitivity = 0.92
false_positive_rate = 0.06

sick = rng.random(n_people) < prevalence
positive_probability = np.where(sick, sensitivity, false_positive_rate)
positive = rng.random(n_people) < positive_probability
print("Odsetek chorych w populacji:", sick.mean())
print("Odsetek dodatnich testów:", positive.mean())
```
