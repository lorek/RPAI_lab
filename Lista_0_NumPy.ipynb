### Wycinek listy a przypisanie
Dla listy `a[:]` tworzy nową listę. Samo `c = a` daje drugą nazwę tej samej listy.
Przed wykonaniem powiedz, co będzie zawierać `b`, a co `c`. Tutaj elementami są liczby; kopia listy jest płytka — zagnieżdżone obiekty nie są kopiowane.

````python
a = [1, 2, 3, 4, 5]
b = a[:]  # A slice creates a new list.
a[2] = -1
print("a:", a)
print("b:", b)

c = a  # Both names refer to the same list.
a[2] = 555
print("c:", c)
````

### Wycinki tablic NumPy
`X[:, :]` wybiera wszystkie wiersze i kolumny; `X[:2, :]` wybiera dwa pierwsze wiersze.
W przeciwieństwie do wycinka listy podstawowy wycinek tablicy NumPy jest widokiem na jej dane. Zmiana elementu widoku zmienia też oryginał.

````python
X = np.array([[1, 2, 3, 4], [4, 5, 6, 7], [2, 2, 1, 1]])
print("X:\n", X)
Y = X[:, :]  # All rows and columns.
print("Cała tablica:\n", Y)
Y = X[:2, :]  # The first two rows, all columns.
print("Dwa pierwsze wiersze:\n", Y)
Y[0, 0] = 1000
print("Po zmianie Y oryginalna tablica X:\n", X)
Z = X[:2, :].copy()
Z[0, 0] = -100
print("Zmiana kopii Z nie zmienia X:", X[0, 0])
````

## 3. `axis`: co sumujemy?
Każdy wiersz poniżej to jedna seria trzech rzutów.
`axis=1` redukuje kolumny: dostajemy jeden wynik na wiersz.
`axis=0` redukuje wiersze: dostajemy jeden wynik na kolumnę.
Bez `axis` redukujemy całą tablicę.

````python
rolls = np.array([[1, 2, 6], [4, 4, 1]])
print(rolls)
print("Sumy serii:", rolls.sum(axis=1))
print("Sumy kolumn:", rolls.sum(axis=0))
print("Średnia wszystkich rzutów:", rolls.mean())
````

Te same operacje możemy zapisać przez funkcje `np.sum` i `np.mean`.
Dla macierzy 2 × 3: bez `axis` otrzymujemy jedną liczbę; z `axis=0` trzy wyniki, po jednym dla każdej kolumny; z `axis=1` dwa wyniki, po jednym dla każdego wiersza.
Sprawdź ręcznie sumę pierwszej kolumny i średnią pierwszego wiersza.

````python
X = np.array([[1, 2, 3], [4, 5, 6]])
print(X)
print("Suma wszystkich elementów:", np.sum(X))
print("Sumy w kolumnach:", np.sum(X, axis=0))
print("Sumy w wierszach:", np.sum(X, axis=1))
print("Średnia wszystkich elementów:", np.mean(X))
print("Średnie w kolumnach:", np.mean(X, axis=0))
print("Średnie w wierszach:", np.mean(X, axis=1))
````

## 4. Maski logiczne: z pytania robimy warunek
`True` liczy się jako 1, a `False` jako 0. Średnia maski jest więc odsetkiem spełnionych warunków.
Do łączenia warunków dla tablic używamy `&` (i), `|` (lub), `~` (negacja) oraz nawiasów.

````python
x = np.array([1, 6, 3, 6, 2, 5])
is_six = (x == 6)
print(is_six)
print(x[is_six])
print("Liczba szóstek:", is_six.sum())
print("Odsetek szóstek:", is_six.mean())
print("Wyniki od 2 do 4:", x[(x >= 2) & (x <= 4)])
````

### Lista składana: wykonujemy to samo dla kilku wartości
Mamy oceny z krótkiego testu. Chcemy dla każdej możliwej oceny obliczyć jej częstość.
Najpierw dla jednej oceny: `grades == 4` sprawdza każdą ocenę, a `.mean()` oblicza odsetek czwórek.

````python
grades = np.array([2, 3, 4, 4, 5, 3, 4, 5])
possible_grades = np.array([2, 3, 4, 5])
print("Maska dla oceny 4:", grades == 4)
print("Częstość oceny 4:", (grades == 4).mean())
````

Teraz powtarzamy obliczenie dla każdej oceny. Zwykła pętla dopisuje wyniki do listy w kolejności 2, 3, 4, 5.

````python
frequency_list = []
for grade in possible_grades:
    frequency = (grades == grade).mean()
    frequency_list.append(frequency)
print(frequency_list)
````

Tę samą pętlę możemy zapisać krócej jako **listę składaną** (*list comprehension*):

```python
[wyrażenie for element in zbiór]
```

Dla kolejnych wartości `grade` obliczamy wyrażenie `(grades == grade).mean()` i umieszczamy wynik w nowej liście.
Nawiasy kwadratowe tworzą listę Pythona. Zewnętrzne `np.array(...)` zamienia ją w tablicę NumPy.

````python
frequency_list_short = [(grades == grade).mean() for grade in possible_grades]
grade_frequencies = np.array(frequency_list_short)
print("Lista:", frequency_list_short)
print("Tablica:", grade_frequencies)
print("Wymiary:", grade_frequencies.shape)

# The same calculation written in one line.
grade_frequencies = np.array([(grades == grade).mean() for grade in possible_grades])
print(grade_frequencies)
````

Dostajemy `[0.125, 0.25, 0.375, 0.25]`: pierwsza wartość dotyczy oceny 2, druga oceny 3 itd.
To nadal pętla Pythona, tylko krócej zapisana — samo użycie `np.array` nie zamienia jej w operację wektorową.

## 7. Pokaz: deska Galtona — gdzie trafią kulki?
Wyobraźmy sobie deskę z rzędami kołków. Kulka spada z góry, przy każdym kołku skręca w lewo lub w prawo i na dole wpada do jednej z przegródek.
Chcemy sprawdzić, ile kulek zbierze się w poszczególnych przegródkach.

**Nasz model:** mamy 20 poziomów. Na każdym poziomie kulka niezależnie wybiera lewo lub prawo z prawdopodobieństwem 1/2. Pomijamy fizykę zderzeń.
Kodujemy lewo jako 0, prawo jako 1. Każdy wiersz tablicy `turns` opisuje jedną kulkę, każda kolumna jeden poziom.

Przegródki numerujemy od 0 do 20. Numer końcowej przegródki jest liczbą skrętów w prawo, czyli sumą wiersza. Same te sumy wystarczą do narysowania histogramu — nie musimy rysować ruchu kulek.

**Najpierw powiedz:** czy przegródki będą równie pełne? Czy skrajna przegródka będzie tak samo prawdopodobna jak środkowa?

````python
n_balls = 10_000
n_levels = 20
turns = rng.integers(0, 2, size=(n_balls, n_levels))
final_bins = turns.sum(axis=1)

plt.figure(figsize=(8, 4))
plt.hist(final_bins, bins=np.arange(n_levels + 2) - 0.5, edgecolor="white")
plt.xlabel("Numer przegródki — liczba skrętów w prawo")
plt.ylabel("Liczba kulek")
plt.title("Gdzie trafiły kulki?")
plt.show()
````

**Interpretacja:** do środkowej przegródki prowadzi wiele sekwencji skrętów, a do skrajnej tylko jedna. Wypisz wszystkie drogi dla 3 poziomów. Czy wyjaśniają wysokości słupków?

### Dodatkowy pokaz: możemy też narysować drogi kulek
Histogram odpowiada już na nasze pytanie. Dla ilustracji możemy dodatkowo pokazać drogi kilkunastu kulek.
To opcjonalny pokaz — nie trzeba umieć samodzielnie napisać tego kodu ani analizować wszystkich poleceń rysowania.

Do rysunku zamieniamy 0 i 1 na kroki −1 i +1. `cumsum` dodaje kolejne kroki, dając pozycję kulki po każdym poziomie. Korzystamy z tych samych losowań co w histogramie.

````python
n_shown = min(15, n_balls)
steps = 2 * turns[:n_shown] - 1
paths = np.column_stack([np.zeros(n_shown), steps.cumsum(axis=1)])
levels = np.arange(n_levels + 1)

fig, ax = plt.subplots(figsize=(7, 4))
for level in range(n_levels):
    peg_x = np.arange(-level, level + 1, 2)
    ax.scatter(peg_x, np.full(peg_x.size, level), color="black", s=10)
ax.plot(paths.T, levels, alpha=0.45)
ax.invert_yaxis()
ax.set(xlabel="Pozycja pozioma", ylabel="Poziom", title="Drogi kilku kulek")
plt.tight_layout()
plt.show()
````

## 8. Operacje na macierzach
`*` mnoży element po elemencie, a `@` mnoży macierze. Transpozycja tablicy jednowymiarowej nie robi z niej kolumny.

````python
A = np.array([[1, 2], [3, 4]])
print("Element po elemencie:\n", A * A)
print("Iloczyn macierzy:\n", A @ A)
print("Transpozycja:\n", A.T)
print("Zera:\n", np.zeros((2, 3)))
print("Jedynki:\n", np.ones((2, 3)))
x = np.array([1, 2, 3])
print("Wektor:", x.shape, "Kolumna:", x[:, None].shape)
print("Różne wartości i liczności:", np.unique([1, 1, 2, 3, 3], return_counts=True))
````

### Wyznacznik i macierz odwrotna
Moduł `np.linalg` zawiera narzędzia algebry liniowej. Dla macierzy o niezerowym wyznaczniku istnieje macierz odwrotna. Iloczyn z odwrotnością powinien dać macierz jednostkową, z dokładnością do błędów obliczeń zmiennoprzecinkowych.

````python
A = np.array([[0, 3, 2], [1, 0, 4], [0, 3, 6]])
print("A:\n", A)
print("Wyznacznik:", np.linalg.det(A))
B = np.linalg.inv(A)
print("Macierz odwrotna:\n", B)
print("Iloczyn A @ B:\n", A @ B)
````

### Mini-projekt: liczba π z losowania punktów
Losujemy punkty równomiernie w kwadracie [0,1] × [0,1]. Pole ćwiartki koła o promieniu 1 wynosi π/4, a pole kwadratu 1.
Jaki warunek sprawdza, czy punkt należy do ćwiartki koła? Dlaczego czterokrotność odsetka takich punktów szacuje π?

````python
points = rng.uniform(0, 1, size=(2000, 2))
inside = (points ** 2).sum(axis=1) <= 1
print("Oszacowanie π:", 4 * inside.mean())
t = np.linspace(0, 1, 200)
plt.figure(figsize=(4, 4))
plt.scatter(points[inside, 0], points[inside, 1], s=3, label="W kole")
plt.scatter(points[~inside, 0], points[~inside, 1], s=3, label="Poza kołem")
plt.plot(t, np.sqrt(1 - t ** 2), color="black")
plt.gca().set_aspect("equal")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
````
